```python
import numpy as np
import pandas as pd
# df = pd.read_csv('/Users/huzihao/Downloads/1a_copy.csv', header=None, names=['X', 'Value1','Value2'])
df = pd.read_csv('/Users/huzihao/Downloads/1a_copy.csv', header=None, names=['X', 'Value1'])
# Format and generate LaTeX output for 8 rows, each with 4 pairs of values
latex_output = ""
for i in range(8):  # Iterate in steps of 4 to generate LaTeX rows
    # Get four consecutive rows for this LaTeX row
    latex_row = ""
    for j in range(4):
        # if i + j*8 < len(df):  # Ensure we don't exceed the number of rows in the data
        x, value1 = df.iloc[i + j*8]
        latex_row += f"{x:.2f} & {value1:.3f}& "

    # Remove the trailing '& ' from the last pair and add the LaTeX newline
    latex_output += latex_row.rstrip('& ') + " \\\\ \n"

# Output the latex formatted string
print(latex_output)

```
