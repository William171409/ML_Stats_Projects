```python
'''
Firstly, Generate  n  random samples uniformly from the interval [0, 1]. This involves creating  n  random numbers  x_i  that are uniformly distributed over this interval.
code:
import numpy as np
n = 200
np.random.seed(798)
x = np.random.uniform(0, 1, n)

Secondly.Compute the values of the function  f(x) = \frac{\exp(-x)}{\sqrt[3]{x}}  at each of the randomly generated points  x_i .
code:
def f(x):
    return np.exp(-x) / np.cbrt(x)
f_values = f(x)

Thirdly,Compute the average of the function values. This average provides an estimate of the integral over the interval [0, 1].
code:
integral_estimate = np.mean(f_values)
Since the interval is [0, 1], there’s no need for adjustment.

Then, we could perform multiple trials to improve the accuracy of the estimate, you can perform multiple Monte Carlo trials and average the results. This helps in reducing variance and providing a more robust estimate.
'''
import numpy as np
n = 200
np.random.seed(798)
def f(x):
    return np.exp(-x) / ( np.cbrt(x))
def monte_carlo_integration_uniform(n):
    x = np.random.uniform(0, 1, n) #生成均匀分布的随机样本
    integral = np.mean(f(x))
    return integral
I_estimate = monte_carlo_integration_uniform(n)
print(f"Estimated value of I using Monte Carlo with uniform distribution: {I_estimate}")
```

## 2)

```python
estimates = [monte_carlo_integration_uniform(n) for _ in range(1000)]
mean_estimate = np.mean(estimates)
std_estimate = np.std(estimates)
print(f"Mean estimate of I over 1000 trials: {mean_estimate}")
print(f"Standard deviation of estimates: {std_estimate}")
```
